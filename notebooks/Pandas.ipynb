## Pandas 
Pandas is a powerful data manipulation library in python, widely used for data analysis and data cleaning. It provides two primary data structures: Series and DataFrame. A series is a one-dimensional array-like object, while a DataFrame is a two-dimensional, size-mutable and potentially heterogeneous tabular data structure with lableled axes (rows and column). 

```python
import pandas as pd
import numpy as np


```

```python
## Series 
## A Series is a one-dimensional labeled array capable of holding any data type.
#  It can be thought of as a column in a table or a single column of data.

data = [1, 2, 3, 4, 5]
series = pd.Series(data)
print(series)
print(type(series))
```

```python
## Create a series from a dictionary
data = {'a': 1, 'b': 2, 'c': 3}
series_from_dict = pd.Series(data)
print(series_from_dict)
```

```python
data= [1, 2, 3, 4, 5]
index = ['a', 'b', 'c', 'd', 'e']
pd.Series(data, index=index)
```

```python
## DataFrame - It is a two-dimensional labeled data structure with columns of potentially different types. It can be thought of as a table or a spreadsheet.
# Create a DataFrame from a dictionary of lists

data = {
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Age': [25, 30, 35],
    'City': ['New York', 'Los Angeles', 'Chicago']
}

df = pd.DataFrame(data)
print(df)
print(type(df))
```

## Data Manipulation and ANalysis with Pandas

```python
data = pd.read_csv('../data/marscrater_pds.csv')
data.head(5)
```

```python
data.tail(5)
```

```python
data.describe()
```

```python
data.dtypes
```

```python
### Handling Missing Data

data.isnull().sum()  # Count the number of missing values in each column
data.isnull().any()  # Check if there are any missing values in the DataFrame
```

```python
## How to fill missing values in a DataFrame

# Fill missing values with a specific value
data.fillna(0, inplace=True)

# Fill missing values with the mean of the column
data.fillna(data.mean(), inplace=True)

# Fill missing values with the median of the column
data.fillna(data.median(), inplace=True)

# Fill missing values with the mode of the column
data.fillna(data.mode().iloc[0], inplace=True)
```

```python
## Filling the missing values in a specific column with the mean of that column
data['column_name'].fillna(data['column_name'].mean(), inplace=True)  # Fill missing values in a specific column with the mean
```

```python
## renaming columns

data.rename(columns={'old_column_name': 'new_column_name'}, inplace=True)  # Rename a specific column
```

```python

dff2=data.rename(columns={'CRATER_ID ': 'ID'})  # Rename a specific column
print(dff2.head(5))
```

```python
## Change datatypes
dff2 = dff2.astype({'ID': 'int64'})  # Change the datatype of a specific column
```

```python
dff2['number_layers']=dff2['NUMBER_LAYERS'].apply(lambda x: x*2)

dff2.head(5)
```

```python
## Data Aggregation and Grouping
grouped_data = data.groupby('CRATER_NAME')['DIAM_CIRCLE_IMAGE'].mean()  # Group by 'CRATER_NAME' and calculate the mean of 'DIAM_CIRCLE_KM'

print(grouped_data)
```

```python
grouped_data2 = data.groupby(['CRATER_NAME', 'MORPHOLOGY_EJECTA_1'])['DIAM_CIRCLE_IMAGE'].sum()
print(grouped_data2)
```

```python

grouped_data3 = data.groupby(['CRATER_NAME', 'MORPHOLOGY_EJECTA_1'])['DIAM_CIRCLE_IMAGE'].mean()
print(grouped_data3)
```

```python
## Aggregate Muliple Functions

grouped_data4 = data.groupby('CRATER_NAME')['DIAM_CIRCLE_IMAGE'].agg(['mean', 'sum', 'count'])
print(grouped_data4)
```

```python
## Merging and Joining DataFrames

df1 = pd.DataFrame({
    'ID': [1, 2, 3],
    'Name': ['Alice', 'Bob', 'Charlie']
})

df2 = pd.DataFrame({
    'ID': [1, 2, 4],
    'Age': [25, 30, 35]
})

df2
```

```python
# Merge 
df_merged = pd.merge(df1, df2, on='ID', how='inner')  # Inner join on 'ID'
print(df_merged)
```

```python
df_merged2= pd.merge(df1, df2, on='ID', how='outer')  # Outer join on 'ID'
print(df_merged2)
```

```python
df_merged3= pd.merge(df1, df2, on='ID', how='left')  # Left join on 'ID'
# Left join on 'ID'- Merge the two DataFrames on the 'ID' column using a left join
print(df_merged3)
```

```python
df_merged4= pd.merge(df1, df2, on='ID', how='right')  
# Right join on 'ID'- Merge the two DataFrames on the 'ID' column using a right join
print(df_merged4)
```
