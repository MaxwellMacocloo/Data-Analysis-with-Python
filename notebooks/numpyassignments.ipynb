```python

### Assignment 4: Statistical Operations

#1. Create a NumPy array of shape (5, 5) filled with random integers. 
# Compute the mean, median, standard deviation, and variance of the array.
dat= np.random.randint(1, 21, size=(5, 5))
data_mean = np.mean(dat)
data_median = np.median(dat)
data_std = np.std(dat)
data_var = np.var(dat)

print("Data:")
print(dat)
print("Mean:")
print(data_mean)
print("Median:")
print(data_median)
print("Standard Deviation:")
print(data_std)
print("Variance:")
print(data_var)


# 2. Create a NumPy array of shape (3, 3) with values from 1 to 9. 
# Normalize the array (i.e., scale the values to have a mean of 0 and a standard deviation of 1).

dat2 = np.arange(1, 10).reshape(3, 3)
data_mean = np.mean(dat2)
data_std = np.std(dat2)
new_data = (dat2 - data_mean) / data_std

print("Data:")
print(dat2)
print("Normalized Data:")
print(new_data)

print("Mean of Normalized Data:", np.mean(new_data))
print("Standard Deviation of Normalized Data:", np.std(new_data))
```

```python
### Assignment 5: Broadcasting

#. Create a NumPy array of shape (3, 3) filled with random integers. 
# Add a 1D array of shape (3,) to each row of the 2D array using broadcasting.

# what is broadcasting in numpy?
# Broadcasting is a powerful mechanism in NumPy that allows for arithmetic 
# operations between arrays of different shapes. When performing operations 
# on arrays, NumPy automatically expands the smaller array along the dimensions 
# of the larger array so that they have compatible shapes. This allows for efficient computation without the need for explicit replication of data.

data = np.random.randint(1, 11, size=(3, 3))
print("Data:")
print(data)

# Create a 1D array of shape (3,)
array_1d = np.array([1, 2, 3])

# Add the 1D array to each row of the 2D array using broadcasting
result = data + array_1d # adds the 1D array to each row of the 2D array using broadcasting

print("1D Array:")
print(array_1d)

print("Result:")
print(result)


# 2. Create a NumPy array of shape (4, 4) filled with random integers. 
# Subtract a 1D array of shape (4,) from each column of the 2D array using broadcasting.

data = np.random.randint(1, 11, size=(4, 4))
print("Data:")
print(data)

# Create a 1D array of shape (4,)
array_1d = np.array([1, 2, 3, 4])

print("1D Array:")
print(array_1d)

# Subtract the 1D array from each column of the 2D array using broadcasting
result = data - array_1d # subtracts the 1D array from each column of the 2D array using broadcasting

print("Result:")
print(result)
```

```python
### Assignment 6: Linear Algebra

#1. Create a NumPy array of shape (3, 3) representing a matrix. 
# Compute its determinant, inverse, and eigenvalues.



# Create a NumPy array of shape (3, 3) representing a matrix
matrix = np.array([[1, 4, 3],
              [2, -1, 2],
              [1, 2, 2]])
print("Matrix:")
print(matrix)

# Compute the determinant
determinant = np.linalg.det(matrix)
print("Determinant:", determinant)

# Compute the inverse
inverse = np.linalg.inv(matrix)
print("Inverse:")
print(inverse)

# Compute the eigenvalues
eigenvalues = np.linalg.eigvals(matrix)
print("Eigenvalues:", eigenvalues)

```

```python
### Assignment 7: Advanced Array Manipulation

# 1. Create a NumPy array of shape (3, 3) with values from 1 to 9. 
# Reshape the array to shape (1, 9) and then to shape (9, 1).

dat= np.arange(1, 10).reshape(3, 3) ## Create a 3x3 array with values from 1 to 9

arr1 =dat.reshape(1, 9) ## Reshape the array to shape (1, 9)
arr2 = dat.reshape(9, 1) ## Reshape the array to shape (9, 1)

print("Original Array:")
print(dat)

print("Array reshaped to (1, 9):")
print(arr1)

print("Array reshaped to (9, 1):")
print(arr2)

# 2. Create a NumPy array of shape (5, 5) filled with random integers. 
# Flatten the array and then reshape it back to (5, 5).

data = np.random.randint(1, 21, size=(5, 5)) ## Create a 5x5 array filled with random integers
data_flat= data.flatten() ## Flatten the array
data_reshaped = data_flat.reshape(5, 5) ## Reshape the flattened array back to (5, 5)

print("Original Array:")
print(data)

print("Flattened Array:")
print(data_flat)

print("Array reshaped back to (5, 5):")
print(data_reshaped)
```
