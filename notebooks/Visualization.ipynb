```python
import matplotlib.pyplot as plt



x= [1,2,3,4,5]
y= [2,3,5,7,11]

plt.plot(x, y)

plt.title("Simple Plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.show()
```

```python
## Create a custmized line plot

plt.figure(figsize=(10, 5))
plt.plot(x, y, color='green', marker='o', linestyle='--', linewidth=2, markersize=8)
```

```python
## Multiple plots 
## Sample data 

x = [1, 2, 3, 4, 5]
x1= [2,4,6,8,10]
y1 = [2, 3, 5, 7, 11]
y2 = [1, 4, 6, 8, 10]

plt.figure(figsize=(10, 5))
plt.subplot(2, 2, 1)
plt.plot(x, y1, color='blue', marker='o', linestyle='-', linewidth=2, markersize=8, label='y1')
plt.title("Plot 1")

plt.legend()

plt.subplot(2, 2, 2)
plt.plot(x, y2, color='green', marker='o', linestyle='-', linewidth=2, markersize=8, label='y2')
plt.title("Plot 2")

plt.legend()

plt.subplot(2, 2, 3)
plt.plot(y1, x, color='red', marker='o', linestyle='-', linewidth=2, markersize=8, label='x')
plt.title("Plot 3")
plt.legend()

plt.subplot(2, 2, 4)
plt.plot(x1, y2, color='violet', marker='o', linestyle='-', linewidth=2, markersize=8, label='y2')
plt.title("Plot 4")

plt.legend()
```

```python
## Bar plots 

categories = ['A', 'B', 'C', 'D', 'E']

values = [10, 15, 7, 12, 9]

plt.bar(categories, values, color='blue')
plt.title("Bar Plot")
plt.xlabel("Categories")
plt.ylabel("Values")
plt.show()
plt.barh(categories, values, color='green')
plt.title("Horizontal Bar Plot")
plt.xlabel("Values")

plt.ylabel("Categories")
plt.show()
```

```python
## Histogram
## Histogram is used to represent the distribution of numerical data by 
# dividing the data into intervals (bins) and counting the number of data points that fall into each bin.


## Example 
data = [12, 2, 14, 5, 8, 9, 11, 13, 16, 17, 18, 19, 20, 21, 22]
plt.hist(data, bins=10, color='orange', edgecolor='black')
```

```python
## Lets create a scatter plot

x = [5, 7, 8, 7, 2, 17, 2, 9, 4, 11]
y = [99, 86, 87, 88, 100, 86, 103, 87, 94, 78]

plt.scatter(x, y, color='red', marker='o')
plt.title("Scatter Plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.show()
```

```python
## Pie chart
labels = ['A', 'B', 'C', 'D']
sizes = [75, 30, 20, 25]
explode=(0.2, 0.3, 0.5, 0) ## Explode the each slice 
plt.pie(sizes, labels=labels, autopct='%1.1f%%', startangle=90, shadow=True, explode=explode)
plt.title("Pie Chart")
plt.axis('equal')  # Equal aspect ratio ensures that pie is drawn as a circle.
```

## Data Visualization with Seaborn

```python
import seaborn as sns

```

```python
## Basic Plotting with Seaborn

tips=sns.load_dataset("tips") 
## The load_dataset() function in Seaborn is used to load built-in datasets that come with the library. These datasets are often used for demonstration purposes, tutorials, and examples. The "tips" dataset is one of the built-in datasets in Seaborn, and it contains information about restaurant tips, including variables such as total bill,
tips

```

```python
sns.scatterplot(data=tips, x='total_bill', y='tip')
plt.title("Scatter Plot of Total Bill vs Tip")
plt.show()
```

```python
## Line plot with Seaborn

sns.lineplot(data=tips, x='total_bill', y='tip')
plt.title("Line Plot of Total Bill vs Tip")
plt.show()

sns.lineplot(data=tips, x='size', y='tip')
plt.title("Line Plot of Size vs Tip")
plt.show()
```

```python
### Categorical Plots with Seaborn

sns.catplot(data=tips, x='day', y='total_bill', kind='box')
plt.title("Box Plot of Total Bill by Day")
plt.show()

sns.boxplot(data=tips, x='day', y='total_bill')
plt.title("Box Plot of Total Bill by Day")
plt.show()


sns.catplot(data=tips, x='day', y='total_bill', kind='bar')
plt.title("Box Plot of Total Bill by Day")
plt.show()


sns.barplot(data=tips, x='day', y='total_bill')
plt.title("Box Plot of Total Bill by Day")
plt.show()
```

```python
## Violin Plot: Gives an idea of the distribution of the data and its probability density. It is similar to a box plot, but it also shows the kernel density estimation of the underlying distribution.

sns.violinplot(data=tips, x='day', y='total_bill')
plt.title("Violin Plot of Total Bill by Day")
plt.show()
```

```python
## Histogram with Seaborn
sns.histplot(data=tips, x='total_bill', bins=10, kde=True)
plt.title("Histogram of Total Bill")
plt.show()
```

```python
## KDE Plot 
sns.kdeplot(data=tips, x='total_bill', fill=True)
plt.title("KDE Plot of Total Bill")
plt.show()
```

```python
## Pairplot : Shows the relationship between multiple variables in a dataset.

sns.pairplot(tips, hue='sex', palette='Set1')
plt.title("Pair Plot of Tips Dataset")
plt.show()


sns.pairplot(tips)
plt.title("Pair Plot of Tips Dataset")
plt.show()
```

```python
## Heatmap: A heatmap is a graphical representation of data where individual values are represented as colors in a matrix. It is commonly used to visualize the correlation between variables in a dataset.

corr = tips[['total_bill', 'tip', 'size']].corr()
corr 
```

```python
sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.title("Heatmap of Correlation Matrix")
plt.show()
```
